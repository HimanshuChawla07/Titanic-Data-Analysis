DATA CLEANING

```python
#we filled all the missing values of Age using its median
df.Age= df.Age.fillna(df.Age.median())
```

```python
# Making the values missing in Cabin as Unknown or we can drop the column
df.Cabin = df.Cabin.fillna("Unknown")
```

```python
df.Embarked = df.Embarked.fillna(df.Embarked.mode()[0])
```

FEATURE ENGINEERING

Feature Engineering is the process of creating new features from existing data to improve analysis or machine learning models.

```python
# all the siblings and spouses + parents and children + 1(themselves)
df['FamilySize'] = df['Sibling_Spouse']+df['Parent_Child']+1
```

```python
df.head()
```

```python
#visualize FamilySize
sns.countplot(x='FamilySize',data=df)
plt.show()
```

```python
#Age Groups
bins = [0,12,18,35,60,100]
labels = ['Child','Teenager','Adult','Middle Age','Senior']
df['AgeGroup']= pd.cut(df['Age'],
                       bins=bins,
                       labels=labels,
                       include_lowest=True)
```

```python
df['Title'] =df.Name.str.extract(r",\s*([^\.]+)\.")
df
```

EXPLORATORY DATA ANALYSIS(EDA)

```python
df.head()
```

```python
# Survival Analysis

#survival rate :-
per_survived = df.Survived.value_counts(normalize=True) * 100
print(per_survived)
```

```python
per_survived.index = ["Not Survived","Survived"]
per_survived.plot.bar(
    xlabel= "Survival Status",
    ylabel = "Percentage (%)",
    title = "Overall Survival Rate"
)
```

```python
#Survived by Gender
fig , axes = plt.subplots(1,2)
sns.countplot(data=df,x="Sex",hue="Survived", ax=axes[0])
axes[0].set_title("Survival count by gender")
axes[0].set_ylabel("Number of Passengers")
axes[0].legend(title = "Survived", labels = ["No","Yes"])

sns.barplot(data=df, x="Sex", y="Survived", errorbar=None, ax=axes[1])
axes[1].set_title("Survival Rate by Gender")
axes[1].set_ylabel("Survival rate")
axes[1].bar_label(axes[1].containers[0], fmt="%.2f")

plt.tight_layout()
```

```python
#Survival by Passenger Class

pclass_survived = (df.groupby("Passanger_Class").Survived.value_counts().unstack())

pclass_survived.plot.bar(
    title="Passenger Survival Count by Class",
    xlabel="Class",
    ylabel="Number Of Passengers"
)
plt.legend(["Did Not Survived","Survived"],title="Status")
plt.show()
```

```python
#Survival by Age Group

age_survived = (df.groupby("AgeGroup").Survived.value_counts().unstack())

age_survived.plot.bar(
    title="Passenger Survival Count by Age Group",
    xlabel="Age Group",
    ylabel="Number Of Passengers"
)
plt.legend(["Did Not Survived","Survived"],title="Status")
plt.show()
```

```python
#Survival by Embarkation Port

embarked_survived = (df.groupby("Embarked").Survived.value_counts().unstack())

embarked_survived_columns = ["Did Not Survive","Survived"]
embarked_survived.plot.bar(
    title="Passenger Survival Count by Embarkation Port",
    xlabel="Embarkation Port",
    ylabel="Number Of Passengers"
)
plt.legend(embarked_survived_columns,title="Status")
plt.show()
```

KEY INSIGHTS

* Only about 38% of passengers survived (342 out of 891). (Overall Survival Rate chart)

* Gender made the biggest difference: 74% of women survived, but only 19% of men. (Survival by Gender chart)

* Class mattered too: 63% of 1st class survived, 47% of 2nd class and just 24% of 3rd class. (Survival by Passenger Class chart)

* Most people travelled in 3rd class (491 passengers, about 55%). (Passenger Class Distribution pie chart)

* Children had the best chances at around 58%, while adults were at 35%. (Survival by Age Group chart)

* Most passengers were young adults, with a median age of 28. (Age Distribution histogram and describe() output)

* Fares were very uneven. The average fare was 32, but a few passengers paid over 500, and 1st class paid far more than the rest. (Fare Distribution and average fare by class chart)

* Passengers travelling alone or in big families (5+) did worse than those in small families of 2 to 4. (Family Size chart)

* Cherbourg passengers survived the most (55%) compared to Southampton (34%). (Survival by Embarkation Port chart)

CONCLUSION

Survival on the Titanic was not random. Women, children and 1st class passengers had much better chances, while men and 3rd class passengers had the worst. Gender and class together had the biggest effect, and family size, fare and embarkation port also showed some differences. These results show patterns, not proven causes.
